```python
import pandas as pd

df = pd.read_csv("financial_data.csv")
print(df.head())
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
print(df.duplicated().sum())

df.drop_duplicates(inplace=True)
```

```python
df['Date'] = pd.to_datetime(df['Date'])

df.sort_values('Date', inplace=True)
```

```python
df['SP500_Lag1'] = df['S&P 500'].shift(1)
df['SP500_Lag5'] = df['S&P 500'].shift(5)
df['SP500_Lag10'] = df['S&P 500'].shift(10)
```

```python
df['SP500_Return'] = df['S&P 500'].pct_change()

df['NASDAQ_Return'] = df['NASDAQ'].pct_change()

df['Gold_Return'] = df['Gold'].pct_change()
```

```python
df['MA_7'] = df['S&P 500'].rolling(7).mean()

df['MA_30'] = df['S&P 500'].rolling(30).mean()
```

```python
df['Volatility_30'] = (
    df['SP500_Return']
    .rolling(30)
    .std()
)
```

```python
df.dropna(inplace=True)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

cols = [
    'S&P 500',
    'NASDAQ',
    'Gold',
    'Oil'
]

df[cols] = scaler.fit_transform(df[cols])
```
